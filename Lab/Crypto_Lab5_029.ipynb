```python
# SHIFT CIPHER
# Encrypt and decrypt using a user-given shift value

def shift_cipher(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            if ch.isupper():
                base = ord('A')
            else:
                base = ord('a')
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            result += ch
    return result
text = input("Enter the text: ")
if text.strip() == "":
    print("Error: Text cannot be empty.")
else:
    try:
        shift = int(input("Enter shift value: "))
        if shift < 0:
            print("Error: Shift cannot be negative.")
        else:
            print("\n1. Encrypt")
            print("2. Decrypt")
            choice = input("Enter your choice: ")
            if choice == "1":
                result = shift_cipher(text, shift)
                print("Encrypted text:", result)
            elif choice == "2":
                result = shift_cipher(text, -shift)
                print("Decrypted text:", result)
            else:
                print("Error: Invalid choice.")
    except ValueError:
        print("Error: Shift must be an integer.")
```

```python
# CAESAR CIPHER
# Caesar cipher uses a fixed shift of 3

def caesar_cipher(text, encrypt=True):
    result = ""
    # Encryption uses +3
    # Decryption uses -3
    shift = 3 if encrypt else -3
    for ch in text:
        if ch.isalpha():
            if ch.isupper():
                base = ord('A')
            else:
                base = ord('a')
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            result += ch
    return result
text = input("Enter the text: ")
if text.strip() == "":
    print("Error: Text cannot be empty.")
else:
    print("\n1. Encrypt")
    print("2. Decrypt")
    choice = input("Enter your choice: ")
    if choice == "1":
        print("Encrypted text:", caesar_cipher(text, True))
    elif choice == "2":
        print("Decrypted text:", caesar_cipher(text, False))
    else:
        print("Error: Invalid choice.")
```

```python
# ROT 13
# Encryption and decryption are the same operation

def rot13(text):
    result = ""
    for ch in text:
        if ch.isalpha():
            if ch.isupper():
                base = ord('A')
            else:
                base = ord('a')
            result += chr((ord(ch) - base + 13) % 26 + base)
        else:
            # Keep spaces and symbols unchanged
            result += ch
    return result
text = input("Enter the text: ")
if text.strip() == "":
    print("Error: Text cannot be empty.")
else:
    result = rot13(text)
    print("ROT13 result:", result)
    print("Applying ROT13 again:", rot13(result))
```

```python
# ATBASH CIPHER
# Atbash replaces each letter with its opposite letter

def atbash(text):
    result = ""
    for ch in text:
        if ch.isupper():
            # A -> Z, B -> Y, etc.
            result += chr(ord('Z') - (ord(ch) - ord('A')))
        elif ch.islower():
            # a -> z, b -> y, etc.
            result += chr(ord('z') - (ord(ch) - ord('a')))
        else:
            # Keep spaces, numbers and symbols unchanged
            result += ch
    return result
text = input("Enter the text: ")
if text.strip() == "":
    print("Error: Text cannot be empty.")
else:
    result = atbash(text)
    print("Encrypted text:", result)
    # Atbash is its own decryption
    print("Decrypted text:", atbash(result))
```

```python
# AFFINE CIPHER
# Encryption: E(x) = (a*x + b) % 26
# Decryption: D(x) = a_inverse*(x-b) % 26

# Find GCD
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a
    
# Find modular inverse
def mod_inverse(a):
    for i in range(1, 26):
        if (a * i) % 26 == 1:
            return i
    return None

# Affine encryption
def affine_encrypt(text, a, b):
    result = ""
    for ch in text:
        if ch.isupper():
            x = ord(ch) - ord('A')
            y = (a * x + b) % 26
            result += chr(y + ord('A'))
        elif ch.islower():
            x = ord(ch) - ord('a')
            y = (a * x + b) % 26
            result += chr(y + ord('a'))
        else:
            result += ch
    return result


# Affine decryption
def affine_decrypt(text, a, b):
    result = ""
    # Find inverse of a
    a_inverse = mod_inverse(a)
    if a_inverse is None:
        return "Error: No modular inverse exists."
    for ch in text:
        if ch.isupper():
            y = ord(ch) - ord('A')
            x = (a_inverse * (y - b)) % 26
            result += chr(x + ord('A'))
        elif ch.islower():
            y = ord(ch) - ord('a')
            x = (a_inverse * (y - b)) % 26
            result += chr(x + ord('a'))
        else:
            result += ch
    return result

text = input("Enter the text: ")
if text.strip() == "":
    print("Error: Text cannot be empty.")
else:
    try:
        a = int(input("Enter value of a: "))
        b = int(input("Enter value of b: "))
        # a must be coprime with 26
        if gcd(a, 26) != 1:
            print("Error: 'a' must be coprime with 26.")
        else:
            # Keep b within 0-25
            b = b % 26
            print("\n1. Encrypt")
            print("2. Decrypt")
            choice = input("Enter your choice: ")
            if choice == "1":
                result = affine_encrypt(text, a, b)
                print("Encrypted text:", result)
            elif choice == "2":
                result = affine_decrypt(text, a, b)
                print("Decrypted text:", result)
            else:
                print("Error: Invalid choice.")
    except ValueError:
        print("Error: a and b must be integers.")
```
