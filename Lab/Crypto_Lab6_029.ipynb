```python
# ONE TIME PAD
# Encryption: C = (P + K) % 26
# Decryption: P = (C - K) % 26

def otp_encrypt(text, key):
    result = ""
    for i in range(len(text)):
        if text[i].isalpha():
            # Convert plaintext and key to numbers
            p = ord(text[i].upper()) - ord('A')
            k = ord(key[i].upper()) - ord('A')
            # OTP encryption
            c = (p + k) % 26
            result += chr(c + ord('A'))
    return result

def otp_decrypt(cipher, key):
    result = ""
    for i in range(len(cipher)):
        c = ord(cipher[i].upper()) - ord('A')
        k = ord(key[i].upper()) - ord('A')
        # OTP decryption
        p = (c - k) % 26
        result += chr(p + ord('A'))
    return result

text = input("Enter plaintext: ").replace(" ", "")
key = input("Enter key: ").replace(" ", "")

# Validation
if text == "":
    print("Error: Plaintext cannot be empty.")
elif key == "":
    print("Error: Key cannot be empty.")
elif not text.isalpha() or not key.isalpha():
    print("Error: Use letters only.")
elif len(text) != len(key):
    print("Error: Key must have the same length as plaintext.")
else:
    encrypted = otp_encrypt(text, key)
    print("Encrypted text:", encrypted)
    decrypted = otp_decrypt(encrypted, key)
    print("Decrypted text:", decrypted)
```

```python
# SHIFT CIPHER BRUTE-FORCE ATTACK
# Try all 26 possible shifts

def shift_decrypt(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            # Decrypt using the given shift
            result += chr((ord(ch) - base - shift) % 26 + base)
        else:
            # Keep spaces unchanged
            result += ch
    return result

# Given ciphertext
ciphertext = "MCI RWR WH"

print("Ciphertext:", ciphertext)
print("\nAll possible plaintexts:\n")

# Try all 26 shifts
for shift in range(26):
    plaintext = shift_decrypt(ciphertext, shift)
    print("Shift", shift, ":", plaintext)
```

```python
# RAIL FENCE CIPHER
# Encryption using a given number of rails

def rail_fence_encrypt(text, rails):
    # Create empty rows
    fence = [""] * rails
    row = 0
    direction = 1
    for ch in text:

        # Add character to current rail
        fence[row] += ch

        # Change direction at top and bottom
        if row == 0:
            direction = 1

        elif row == rails - 1:
            direction = -1

        row += direction

    # Combine all rails
    result = ""

    for rail in fence:
        result += rail

    return result


# Get input
text = input("Enter plaintext: ")

if text.strip() == "":
    print("Error: Plaintext cannot be empty.")

else:
    try:
        rails = int(input("Enter number of rails: "))

        # Validation
        if rails < 2:
            print("Error: Number of rails must be at least 2.")

        elif rails > len(text):
            print("Error: Number of rails cannot be greater than text length.")

        else:
            encrypted = rail_fence_encrypt(text, rails)

            print("Encrypted text:", encrypted)

    except ValueError:
        print("Error: Number of rails must be an integer.")
```
