```python
#16. Character frequency
text = "CRYPTOLOGYISINTERESTING"

frequency = {}

for character in text:
    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

print("Character Frequency:")

for character in frequency:
    print(character, ":", frequency[character])

most_frequent = max(frequency, key=frequency.get)

print("Most frequently occurring character:", most_frequent)
```

```python
#17. Percentage frequency

text = "CRYPTOLOGYISINTERESTING"

frequency = {}

for character in text:
    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

total = len(text)

for character in frequency:
    percentage = (frequency[character] / total) * 100
    print(character, ":", round(percentage, 2), "%")
```

```python
#20. Prime numbers
#Check whether a number is prime
num = int(input("Enter a number: "))
if num < 2:
    print("Not a prime number")
else:
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("Prime number")
    else:
        print("Not a prime number")


#Display prime numbers within a range
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
print("Prime numbers:")

for num in range(start, end + 1):
    if num >= 2:
        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            print(num)
```

```python
#21. GCD and coprime numbers
import math

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
gcd = math.gcd(a, b)
print("GCD:", gcd)

if gcd == 1:
    print("The numbers are coprime")
else:
    print("The numbers are not coprime")
```

```python
#22. Multiplicative inverse of an integer b
b = int(input("Enter b: "))
inverse = None
for x in range(1, 26):
    if (b * x) % 26 == 1:
        inverse = x
        break

if inverse is not None:
    print("Multiplicative inverse:", inverse)
else:
    print("Multiplicative inverse does not exist")
```

```python
#23. Client and Server Communication
import socket

server = socket.socket()
server.bind(("localhost", 9999))
server.listen(1)
print("Waiting for client...")

connection, address = server.accept()
print("Client connected")

message = connection.recv(1024).decode()
print("Message from client:", message)

connection.send("Hello from server".encode())

connection.close()
server.close()
```
