```python
#3. Read cryptology.txt one line at a time and display each line with its line number
file = open("cryptology.txt", "r")

line_number = 1
for line in file:
    print(line_number, line.strip())
    line_number = line_number + 1

file.close()
```

```python
#4. Store the output in cryptology_no.txt
file = open("cryptology.txt", "r")
output = open("cryptology_no.txt", "w")
line_number = 1
for line in file:
    result = str(line_number) + " " + line.strip()
    print(result)
    output.write(result + "\n")
    line_number = line_number + 1

file.close()
output.close()

print("Output stored in cryptology_no.txt")
```

```python
#5. Read cryptology.txt and determine the number of lines, words, and characters
file = open("cryptology.txt", "r")

content = file.read()

lines = content.splitlines()
words = content.split()
characters = len(content)

print("Number of lines:", len(lines))
print("Number of words:", len(words))
print("Number of characters:", characters)

file.close()
```

```python
#6. Read cryptology_no.txt and count uppercase, lowercase, digits, spaces, and special characters
file = open("cryptology_no.txt", "r")

content = file.read()

uppercase = 0
lowercase = 0
digits = 0
spaces = 0
special = 0

for ch in content:
    if ch.isupper():
        uppercase = uppercase + 1
    elif ch.islower():
        lowercase = lowercase + 1
    elif ch.isdigit():
        digits = digits + 1
    elif ch == " ":
        spaces = spaces + 1
    elif ch != "\n":
        special = special + 1

print("Uppercase letters:", uppercase)
print("Lowercase letters:", lowercase)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)

file.close()
```

```python
#7. Accept a word from the user and search for it in cryptology.txt
file = open("cryptology.txt", "r")
word = input("Enter the word to search: ")
found = False

for line in file:
    if word in line:
        print("Word found:", line.strip())
        found = True

if found == False:
    print("Word not found.")

file.close()
```

```python
#9. Convert all alphabetic characters to uppercase and store in normalized.txt
file = open("cryptology.txt", "r")
output = open("normalized.txt", "w")

for line in file:
    line = line.upper()
    output.write(line)

file.close()
output.close()

print("Contents stored in normalized.txt")
```

```python
#10. Read cryptology_no.txt and print only lines containing only alphabetic characters
file = open("cryptology_no.txt", "r")

for line in file:
    line = line.strip()

    if any(ch.isalpha() for ch in line):
        print(line)

file.close()
```

```python
#Combined Regex Program
import re

file = open("security_data.txt", "r")
output = open("pattern_results.txt", "w")

content = file.read()

# Find IP addresses
ip_addresses = re.findall(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', content)

# Find email addresses
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', content)

# Find session IDs
session_ids = re.findall(r'SEC-\d{4}-\d{4}', content)

# Find hexadecimal sequences
hex_values = re.findall(r'\b[0-9a-fA-F]{6,}\b', content)

output.write("IP Addresses:\n")
for ip in ip_addresses:
    output.write(ip + "\n")

output.write("\nEmail Addresses:\n")
for email in emails:
    output.write(email + "\n")

output.write("\nSession IDs:\n")
for session in session_ids:
    output.write(session + "\n")

output.write("\nHexadecimal Sequences:\n")
for value in hex_values:
    output.write(value + "\n")

file.close()
output.close()

print("Results stored in pattern_results.txt")
file = open("pattern_results.txt", "r")

print(file.read())

file.close()
```
